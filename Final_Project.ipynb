```python
class TDManagement:
    def __init__(self):
        self.vehicles = []

    def insert_vehicle(self, v_id, v_type, speed):
        vehicle = {"id": v_id, "type": v_type, "speed": speed}
        self.vehicles.append(vehicle)

    def analyze_traffic(self):
        C = 0
        T = 0
        MC = 0

        for v in self.vehicles:
            if v["type"] == "Car":
                C += 1
            elif v["type"] == "Truck":
                T += 1
            elif v["type"] == "Motorcycle":
                MC += 1

        total = len(self.vehicles)
        print(f"Total: {total}")

        if total > 0:
            print(f"Car: {C} = {round(C / total * 100, 1)}%")
            print(f"Truck: {T} = {round(T / total * 100, 1)}%")
            print(f"Motorcycle: {MC} = {round(MC / total * 100, 1)}%")
        print("-----------")

    def delete_vehicle(self, v_id):
        for i in range(len(self.vehicles)):
            if self.vehicles[i]["id"] == v_id:
                self.vehicles.pop(i)
                break

system = TDManagement()

print("SAMPLE 1: Normal Traffic")
system.insert_vehicle("V01", "Car", 45)
system.insert_vehicle("V02", "Motorcycle", 50)
system.insert_vehicle("V03", "Car", 60)
system.insert_vehicle("V04", "Truck", 40)
system.insert_vehicle("V05", "Car", 55)
system.insert_vehicle("V06", "Motorcycle", 45)
system.insert_vehicle("V07", "Car", 50)
system.insert_vehicle("V08", "Truck", 35)
system.insert_vehicle("V09", "Car", 60)
system.insert_vehicle("V10", "Motorcycle", 55)
system.analyze_traffic()

for i in range(1, 11):
    system.delete_vehicle("V" + str(i).zfill(2))


print("SAMPLE 2: Peak Hour")
for i in range(11, 26):
    if i % 5 == 0:
        system.insert_vehicle(f"V{i}", "Motorcycle", 15)
    else:
        system.insert_vehicle(f"V{i}", "Car", 15)
system.analyze_traffic()

for i in range(11, 26):
    system.delete_vehicle(f"V{i}")


print("SAMPLE 3: Midnight Logistics")
for i in range(26, 38):
    if i <= 35:
        system.insert_vehicle(f"V{i}", "Truck", 50)
    else:
        system.insert_vehicle(f"V{i}", "Car", 50)
system.analyze_traffic()

for i in range(26, 38):
    system.delete_vehicle(f"V{i}")


print("SAMPLE 4: Motorway Event")
for i in range(38, 48):
    system.insert_vehicle(f"V{i}", "Car", 85)
for i in range(48, 58):
    system.insert_vehicle(f"V{i}", "Truck", 80)
system.analyze_traffic()

for i in range(38, 58):
    system.delete_vehicle(f"V{i}")
```
